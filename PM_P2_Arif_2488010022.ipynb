#1. Apa perbedaan supervised, unsupervised, dan reinforcement learning?
Jadi ringkasan perbedaan dari ke-3 jenis learning tersebut adalah
-Supervised learning menggunakan data yang memiliki label/jawaban
-Unsupervised learning menggunakan data tanpa label dan mencari pola sendiri/segmentasi
-Reinforcement learning belajar melalui percobaan, tindakan, reward, dan punishment

#2. Mengapa pembagian data latih dan uji penting?
Pembagian data menjadi data latih (training) dan data uji (testing) penting untuk mengetahui apakah model dapat bekerja dengan baik pada data baru, bukan hanya menghafal data yang sudah dipelajari. Jika model hanya diuji menggunakan data yang sudah digunakan untuk belajar, hasilnya bisa terlihat bagus tetapi belum tentu mampu memprediksi data baru.
